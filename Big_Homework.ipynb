````python
def input_fio_list():
    """
    Функция для ввода ФИО до слова "Стоп" (регистр слова не важен)
    Возвращает список введённых строк
    """
    fio_list = []
    s = input()
    while s.lower() != 'стоп':
        fio_list.append(s)
        s = input()
   
    
    return fio_list


def capitalize_fio(fio_list):
    """
    Функция для преобразования регистра ФИО 
    Принимает список ФИО, возвращает СПИСОК с преобразованным регистром
    """
    capitalized_list = list(map(str.lower, fio_list))
    capitalized_list = list(map(str.title, fio_list))
    
    # Ваш код здесь для подпункта b)
    return capitalized_list


def format_fio(fio_list):
    """
    Функция для форматирования ФИО: фамилия полностью, инициалы с точками
    Принимает список ФИО (уже с большими буквами), возвращает ОДНУ строку через запятую
    """
    new_fio_list = []
    for tmp_list in fio_list:
        fio_tmp_list = tmp_list.split()
        fio_tmp_str = ' '.join([fio_tmp_list[i]  if i == 0 else fio_tmp_list[i][0] + '.' for i in range(len(fio_tmp_list))])
        new_fio_list.append(fio_tmp_str)
    formatted_string = ", ".join(new_fio_list)
    # Ваш код здесь для подпункта c)
    
    return formatted_string
````

````python
# Запуск: вводите ФИО по одному в строке, в конце введите «Стоп»

# Ввод данных
fio_list = input_fio_list()

# Преобразование регистра
capitalized_fio = capitalize_fio(fio_list)

# Форматирование ФИО
result = format_fio(capitalized_fio)

# Вывод результата
print(result)
````
